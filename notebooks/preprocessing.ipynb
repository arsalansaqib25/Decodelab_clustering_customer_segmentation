# Feature Engineering

```python
reference_date = purchase_df["InvoiceDate"].max() + pd.Timedelta(days=1)
```

```python
recency = purchase_df.groupby("CustomerID").agg(
    Recency=("InvoiceDate", lambda x:
             (reference_date - x.max()).days)
)
```

```python
frequency = purchase_df.groupby("CustomerID").agg(
    Frequency=("InvoiceNo", "nunique")
)
```

```python
monetary = purchase_df.groupby("CustomerID").agg(
    Monetary=("TotalAmount", "sum")
)
```

```python
customer_features = purchase_df.groupby("CustomerID").agg(
    
    # Monetary features
    Total_Spend=("TotalAmount", "sum"),
    Average_Order_Value=("TotalAmount", "mean"),
    Median_Order_Value=("TotalAmount", "median"),
    Maximum_Order_Value=("TotalAmount", "max"),
    Minimum_Order_Value=("TotalAmount", "min"),
    
    # Quantity features
    Total_Quantity=("Quantity", "sum"),
    Average_Quantity=("Quantity", "mean"),
    Maximum_Quantity=("Quantity", "max"),
    Minimum_Quantity=("Quantity", "min"),
    
    # Transaction features
    Total_Transactions=("InvoiceNo", "nunique"),
    Unique_Products=("StockCode", "nunique"),
    Unique_Descriptions=("Description", "nunique"),
    
    # Price features
    Average_Unit_Price=("UnitPrice", "mean"),
    Maximum_Unit_Price=("UnitPrice", "max"),
    Minimum_Unit_Price=("UnitPrice", "min"),
    
    # Geographic behavior
    Countries=("Country", "nunique"),
    
    # Temporal behavior
    First_Purchase=("InvoiceDate", "min"),
    Last_Purchase=("InvoiceDate", "max"),
    
).reset_index()
```

```python
# Recency -> How recenctly did the customer purchase
customer_features["Recency"] = (
    reference_date -
    customer_features["Last_Purchase"]
).dt.days
```

```python
# Customer lifetime
customer_features["Customer_Lifetime"] = (
    customer_features["Last_Purchase"] -
    customer_features["First_Purchase"]
).dt.days
```

```python
# Spend per transaction
customer_features["Spend_per_Transaction"] = (
    customer_features["Total_Spend"] /
    customer_features["Total_Transactions"]
)
```

```python
# Quantity per transaction
customer_features["Quantity_per_Transaction"] = (
    customer_features["Total_Quantity"] /
    customer_features["Total_Transactions"]
)
```

```python
# Product diversity
customer_features["Product_Diversity"] = (
    customer_features["Unique_Products"] /
    customer_features["Total_Transactions"]
)
```

```python
# Cancellation features
cancellation_features = df.groupby("CustomerID").agg(
    Total_Invoices=("InvoiceNo", "nunique"),
    Cancelled_Transactions=("IsCancelled", "sum")
)
```

```python
# Cancellation rate
cancellation_features["Cancellation_Rate"] = (
    cancellation_features["Cancelled_Transactions"] /
    cancellation_features["Total_Invoices"]
)
```

```python
# Merging two dataset
customer_features = customer_features.merge(
    cancellation_features[["Cancellation_Rate"]],
    on="CustomerID",
    how="left"
)
```

```python
#Fill nane values of Canellaion rate with 0
customer_features["Cancellation_Rate"] = (
    customer_features["Cancellation_Rate"].fillna(0)
)
```

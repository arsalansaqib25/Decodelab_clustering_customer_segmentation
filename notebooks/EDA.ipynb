# About the data
- Total_Spend -> Total Revenue/ How much money does the customer spend?
- Average_Order_Value -> Average Revenue
- Median_Order_Value -> Median Revenue
- Maximum_Order_Value -> Maximum Revenue
- Minimum_Order_Value -> Minimum Revenue
- Total_Quantity -> Total number of items purchased by the customer.
- Average_Quantity ->	Average number of items purchased per transaction/record.
- Maximum_Quantity ->	Largest quantity purchased in a single transaction/record.
- Minimum_Quantity	-> Smallest quantity purchased in a single transaction/record.
- Total_Transactions	-> Number of unique invoices/orders made by the customer.
- Unique_Products	-> Number of different product codes purchased by the customer.
- Unique_Descriptions ->	Number of different product descriptions purchased
- Average_Unit_Price	-> Average price per individual item purchased.
- Maximum_Unit_Price	-> Highest unit price among the customer's purchases.
- Minimum_Unit_Price ->	Lowest unit price among the customer's purchases.
- Recency -> How recenctly did the customer purchase
- Countries ->      Number of different countries associated with the customer's transactions.

```python
log_features = [
    "Minimum_Quantity",
    "Minimum_Order_Value",
    "Median_Order_Value",
    "Average_Quantity",
    "Minimum_Unit_Price",
    "Maximum_Order_Value",
    "Average_Unit_Price",
    "Quantity_per_Transaction",
    "Average_Order_Value",
    "Maximum_Quantity",
    "Spend_per_Transaction",
    "Maximum_Unit_Price",
    "Countries",
    "Total_Quantity",
    "Total_Spend",
    "Total_Transactions",
    "Unique_Descriptions",
    "Unique_Products",
    "Recency"
]

for col in log_features:
    df[col] = np.log1p(df[col])
```
